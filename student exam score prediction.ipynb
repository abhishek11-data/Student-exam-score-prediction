```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("D:\student habbit performance\student_habits_performance.csv")
df
```

```python
df.info()
```

```python
df.drop(["student_id"],axis=1, inplace=True)
df
```

```python
df.isnull().sum()
```

```python
df["parental_education_level"].unique()
```

```python
df["parental_education_level"]=df["parental_education_level"].fillna(df["parental_education_level"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df["gender"]=le.fit_transform(df["gender"])
df["part_time_job"]=le.fit_transform(df["part_time_job"])
df["diet_quality"]=le.fit_transform(df["diet_quality"])
df["parental_education_level"]=le.fit_transform(df["parental_education_level"])
df["internet_quality"]=le.fit_transform(df["internet_quality"])
df["extracurricular_participation"]=le.fit_transform(df["extracurricular_participation"])
```

```python
x=df.iloc[:,:-1]
x
```

```python
y=df.iloc[:,-1]
y
```

```python
df
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
x_train.shape
```

```python
y_test.shape
```

LINEAR REGRESSOR


```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
y_pred=lr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.metrics import root_mean_squared_error
```

```python
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
print(root_mean_squared_error(y_test,y_pred))
```

RANDOM FOREST REGRESSOR


```python
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor()
```

```python
rfr.fit(x_train,y_train)
y_pred=rfr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.metrics import root_mean_squared_error
```

```python
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
print(root_mean_squared_error(y_test,y_pred))
```

DECISION TREE REGRESSOR


```python
from sklearn.tree import DecisionTreeRegressor
dtr=DecisionTreeRegressor()
```

```python
dtr.fit(x_train,y_train)
y_pred=dtr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.metrics import root_mean_squared_error
```

```python
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
print(root_mean_squared_error(y_test,y_pred))
```

RANDOM FOREST RESGRESSOR + GRIDSEARCH CV


```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GridSearchCV
```

```python
rfr=RandomForestRegressor()
```

```python
rfr.fit(x_train,y_train)
```

```python
param_grid= {
    'n_estimators':[50,100,200],
    'max_depth':[5,10,20, None],
    'min_samples_split':[2,5,10],
    'min_samples_leaf':[1,2,4]
}
```

```python
rfr_grid=GridSearchCV(estimator=rfr,param_grid=param_grid,cv=5,scoring='r2',n_jobs=-1)
```

```python
rfr_grid.fit(x_train,y_train)
```

```python
print(rfr_grid.best_params_)
```

```python
print(rfr_grid.best_score_)
```

```python
best_rfr=rfr_grid.best_estimator_
```

```python
y_pred=best_rfr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import root_mean_squared_error
from sklearn.metrics import r2_score
```

```python
print(mean_absolute_error(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
print(root_mean_squared_error(y_test,y_pred))
print(r2_score(y_test,y_pred))
```

DECISIONTREEREGRESSOR + GRIDSEARCH CV

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import GridSearchCV
```

```python
dtr=DecisionTreeRegressor()
```

```python
dtr.fit(x_train,y_train)
```
